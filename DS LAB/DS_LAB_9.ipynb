```python
import numpy as np
from scipy.stats import bernoulli, binom

# Bernoulli Distribution
p = 0.7

print("----- Bernoulli Distribution -----")

samples = bernoulli.rvs(p, size=10)
print("Student Outcomes:", samples)
print("Probability of Pass:", bernoulli.pmf(1, p))
print("Probability of Fail:", bernoulli.pmf(0, p))

# Binomial Distribution
print("\n----- Binomial Distribution -----")

n = 10
k = 8

probability = binom.pmf(k, n, p)

print("Number of Students:", n)
print("Probability of Passing:", p)
print("Probability that exactly 8 students pass:", probability)

random_binomial = binom.rvs(n, p, size=5)
print("Random Binomial Outcomes:", random_binomial)
```

```python
import numpy as np
from scipy import stats
from statsmodels.stats.weightstats import ztest

# Z-Test
print("----- Z-Test -----")

sample_marks = np.array([
    78, 82, 85, 88, 90,
    76, 95, 89, 84, 91
])

population_mean = 80

z_stat, p_value = ztest(
    sample_marks,
    value=population_mean
)

print("Sample Marks:", sample_marks)
print("Population Mean:", population_mean)
print("Z-Statistic:", z_stat)
print("P-Value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")

# Independent t-Test
print("\n----- Independent t-Test -----")

class_A = np.array([78, 82, 85, 88, 90])
class_B = np.array([72, 75, 70, 68, 74])

t_stat, p_value = stats.ttest_ind(
    class_A, class_B
)

print("Class A Marks:", class_A)
print("Class B Marks:", class_B)
print("t-Statistic:", t_stat)
print("P-Value:", p_value)

if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```
