```python
import pandas as pd
import numpy as np

# Create Dataset with Missing Values
df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 35],
    'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']
})

# Display Original Dataset
print("Original Dataset (with Missing Values):")
print(df)

# Mean Imputation for Numeric Column
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Mode Imputation for Categorical Column
df['Department'] = df['Department'].fillna(
    df['Department'].mode()[0]
)

# Display Dataset after Imputation
print("\nDataset after Handling Missing Values:")
print(df)
```

```python
import pandas as pd
import numpy as np

# Create Dataset with Missing Values
df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 35],
    'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']
})

# Display Original Dataset
print("Original Dataset (with Missing Values):")
print(df)

#forward fill
df_ffill=df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)
#backward fill
df_bfill=df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
import numpy as np
df=pd.DataFrame({
    'Age':[25,30,np.nan,40,35],
    'Department':['HR','Finance','Finance',np.nan,'IT']
})
#Display orignial dataset
print("Original Dataset (with Missing Values):")
print(df)

#Drop row with missing values
df_drop_rows=df.dropna()
print("After dropping rows:\n",df_drop_rows)
#Drop column with missing values
df_drop_cols=df.dropna(axis=1)
print("After dropping columns:\n",df_drop_cols)
```

```python
import pandas as pd

# Sample dataset with duplicates
df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 4, 5],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David', 'David'],
    'Age': [25, 30, 30, 35, 40, 40, 40]
})
# Display original data
print("Original Data:\n",df)
# Remove duplicates based only on ID
df_subset_id = df.drop_duplicates(subset=['ID'])
print("\nAfter Subset-Based Removal (ID):\n",df_subset_id)
# Remove duplicates based only on Name
df_subset_name = df.drop_duplicates(subset=['Name'])
print("\nAfter Subset-Based Removal (Name):\n",df_subset_name)
```

```python

import pandas as pd
df=pd.DataFrame({
    'Date':['2025-01-05','05?01?2025','jan 5,2025','2025.01.05']
})
print("Original Data:\n",df)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce').dt.strftime('%Y-%m-%d')

print(df)
```

```python
import pandas as pd

# Sample dataset with inconsistent text case
df = pd.DataFrame({
    'Name': ['Alice', 'BOB', 'Charlie', 'david']
})

# Display original data
print("Original Data:\n")
print(df)

# Convert all names to lowercase
df['Name_lower'] = df['Name'].str.lower()

# Convert all names to uppercase
df['Name_upper'] = df['Name'].str.upper()

# Display the final data
print("\nAfter Case Normalization:\n")
print("\n",df)
```
